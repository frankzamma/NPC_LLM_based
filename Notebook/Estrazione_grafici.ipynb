```python
#Architettura 2
def modifica_consigli_accettati (directory):
  consigli_accettati_csv = pd.read_csv(directory + "csv_consigli_accettati_Gemini.csv")

  steps_csv = pd.read_csv(directory + "csv_steps_Gemini.csv")

  steps_dict = dict(zip(steps_csv["Episode"], steps_csv["Steps"]))

  for i in range(len(consigli_accettati_csv)):
    episode = consigli_accettati_csv.loc[i, "Episode"]
    if episode in steps_dict:
      consigli_accettati_csv.loc[i, "Consigli_Accettati"] = steps_dict[episode] - consigli_accettati_csv.loc[i, "Consigli_Accettati"]

  consigli_accettati_csv.to_csv("/content/csv_consigli_accettati_Gemini.csv", index=False)
```

```python
def mostra_grafici (directory, xlabel, ylabel, title):

  df = pd.read_csv(directory)

  plt.figure(figsize=(12, 6))
  plt.plot(df[df.columns[0]], df[df.columns[1]], color="#6495ed")

  # Etichette e titolo
  plt.xlabel(xlabel)
  plt.ylabel(ylabel)
  plt.title(title)
  plt.legend()
  plt.grid(True)

  # Mostra il grafico
  plt.show()
```

```python
def mostra_frequenza_relativa (directory, xlabel, ylabel, title):
  df = pd.read_csv(directory)

  frequenza_relativa = df[df.columns[1]].value_counts(normalize=True)

  # Creazione del grafico a barre
  plt.figure(figsize=(8, 5))
  plt.bar(frequenza_relativa.index, frequenza_relativa.values, color="skyblue")

  # Etichette e titolo
  plt.xlabel(xlabel)
  plt.ylabel(ylabel)
  plt.title(title)

  # Mostra il grafico
  plt.show()
```
